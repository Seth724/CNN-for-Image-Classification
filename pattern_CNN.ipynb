```python
import json
from pathlib import Path
import pandas as pd

root = Path("/content/drive/MyDrive/pattern_cnn_experiments")

rows = []

for evaluation_file in root.glob("*/metrics/evaluation.json"):
    with evaluation_file.open() as f:
        metrics = json.load(f)

    rows.append({
        "experiment": evaluation_file.parent.parent.name,
        "test_accuracy": metrics.get("test_accuracy"),
        "macro_precision": metrics.get("macro_precision"),
        "macro_recall": metrics.get("macro_recall"),
        "macro_f1": metrics.get("macro_f1"),
        "weighted_f1": metrics.get("weighted_f1"),
        "test_loss": metrics.get("test_loss"),
        "epochs": metrics.get("epochs"),
        "batch_size": metrics.get("batch_size"),
        "optimizer": metrics.get("optimizer"),
        "learning_rate": metrics.get("learning_rate"),
        "training_seconds": metrics.get("training_seconds"),
    })

comparison = pd.DataFrame(rows)

if not comparison.empty:
    comparison = comparison.sort_values(
        by=["test_accuracy", "macro_f1"],
        ascending=False,
    )

display(comparison)
comparison.to_csv(root / "model_a_comparison.csv", index=False)
```
