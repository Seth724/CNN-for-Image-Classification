```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

root = Path("/content/drive/MyDrive/pattern_cnn_experiments")

selected_runs = [
    "model_a_rmsprop_lr0.002_bs128_e20",
    "model_a_adam_lr0.003_bs64_e20",
]

class_names = [
    "T-shirt/top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle boot",
]

for run_name in selected_runs:
    run_dir = root / run_name
    metrics_dir = run_dir / "metrics"
    report_dir = run_dir / "report_figures"
    report_dir.mkdir(parents=True, exist_ok=True)

    history = pd.read_csv(metrics_dir / "history.csv")

    with (metrics_dir / "evaluation.json").open() as file:
        metrics = json.load(file)

    # Loss curves
    plt.figure(figsize=(8, 5))
    plt.plot(history["epoch"], history["loss"], label="Training loss", linewidth=2)
    plt.plot(
        history["epoch"],
        history["val_loss"],
        label="Validation loss",
        linewidth=2,
    )
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(f"{run_name}: loss curves")
    plt.grid(alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(report_dir / "loss_curves.png", dpi=300)
    plt.close()

    # Accuracy curves
    plt.figure(figsize=(8, 5))
    plt.plot(
        history["epoch"],
        history["accuracy"],
        label="Training accuracy",
        linewidth=2,
    )
    plt.plot(
        history["epoch"],
        history["val_accuracy"],
        label="Validation accuracy",
        linewidth=2,
    )
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.title(f"{run_name}: accuracy curves")
    plt.grid(alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(report_dir / "accuracy_curves.png", dpi=300)
    plt.close()

    # Confusion matrix
    confusion = np.array(metrics["confusion_matrix"])

    plt.figure(figsize=(9, 7))
    sns.heatmap(
        confusion,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=class_names,
        yticklabels=class_names,
    )
    plt.xlabel("Predicted label")
    plt.ylabel("True label")
    plt.title(f"{run_name}: test confusion matrix")
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.savefig(report_dir / "confusion_matrix.png", dpi=300)
    plt.close()

    # Per-class precision, recall, and F1
    per_class = metrics.get("per_class")

    if per_class:
        if isinstance(per_class, list):
            per_class_df = pd.DataFrame(per_class)
            per_class_df["class"] = class_names
        else:
            per_class_df = pd.DataFrame(per_class).T
            per_class_df["class"] = per_class_df.index

        metric_columns = [
            column
            for column in ["precision", "recall", "f1_score", "f1"]
            if column in per_class_df.columns
        ]

        if metric_columns:
            per_class_df.set_index("class")[metric_columns].plot(
                kind="bar",
                figsize=(12, 6),
                ylim=(0, 1),
            )
            plt.xlabel("Class")
            plt.ylabel("Score")
            plt.title(f"{run_name}: per-class metrics")
            plt.xticks(rotation=45, ha="right")
            plt.grid(axis="y", alpha=0.3)
            plt.tight_layout()
            plt.savefig(report_dir / "per_class_metrics.png", dpi=300)
            plt.close()

    print("Created figures:", report_dir)
```

```python
import json
from pathlib import Path

root = Path("/content/drive/MyDrive/pattern_cnn_experiments")

selected_runs = [
    "model_a_rmsprop_lr0.002_bs128_e20",
    "model_a_adam_lr0.003_bs64_e20",
]

for run_name in selected_runs:
    run_dir = root / run_name
    evaluation_path = run_dir / "metrics" / "evaluation.json"
    architecture_path = run_dir / "metrics" / "architecture.json"

    with evaluation_path.open() as file:
        evaluation = json.load(file)

    with architecture_path.open() as file:
        architecture = json.load(file)

    evaluation["total_parameters"] = architecture["total_parameters"]

    evaluation_path.write_text(
        json.dumps(evaluation, indent=2)
    )

    print(run_name)
    print("Total parameters:", evaluation["total_parameters"])
```

```python
import json
from pathlib import Path
import pandas as pd

root = Path("/content/drive/MyDrive/pattern_cnn_experiments")

selected_runs = [
    "model_a_rmsprop_lr0.002_bs128_e20",
    "model_a_adam_lr0.003_bs64_e20",
]

rows = []

for run_name in selected_runs:
    evaluation_path = root / run_name / "metrics" / "evaluation.json"

    with evaluation_path.open() as file:
        metrics = json.load(file)

    rows.append({
        "model": "Model A",
        "run": run_name,
        "optimizer": metrics.get("optimizer"),
        "learning_rate": metrics.get("learning_rate"),
        "epochs": metrics.get("epochs"),
        "batch_size": metrics.get("batch_size"),
        "total_parameters": metrics.get("total_parameters"),
        "model_size_kb": metrics.get("model_size_kb"),
        "training_time_per_epoch_s": metrics.get("mean_epoch_time_s"),
        "test_accuracy": metrics.get("test_accuracy"),
        "macro_precision": metrics.get("macro_precision"),
        "macro_recall": metrics.get("macro_recall"),
        "macro_f1": metrics.get("macro_f1"),
    })

model_a_comparison = pd.DataFrame(rows)

model_a_comparison = model_a_comparison.sort_values(
    "test_accuracy",
    ascending=False,
)

display(model_a_comparison)

model_a_comparison.to_csv(
    root / "model_a_comparison.csv",
    index=False,
)
```
